Question 1 : What is a RESTful API?

Answer:  A RESTful API (Representational State Transfer API) is a web service that follows REST principles, allowing communication between client and server using standard HTTP methods like GET, POST, PUT, and DELETE. It uses resources identified by URLs and typically exchanges data in formats such as JSON or XML, making it lightweight, scalable, and easy to integrate across different platforms. RESTful APIs are stateless, meaning each request from client to server must contain all the information needed to understand and process the request. They support CRUD operations (Create, Read, Update, Delete) through HTTP methods, enabling efficient resource manipulation. Additionally, RESTful APIs promote a uniform interface, which simplifies and decouples the architecture, improving scalability and flexibility. They are widely used in web and mobile applications due to their simplicity, performance, and compatibility with various data formats and platforms.

Question 2: What is Flask, and why is it popular for building APIs?

Answer: Flask is a lightweight Python web framework designed for building web applications and APIs. It is popular because it is simple, flexible, and easy to learn, making it ideal for rapid development. Question 3: What are HTTP methods used in RESTful APIs? (20 marks) Answer:
The HTTP methods used in RESTful APIs are essential for performing CRUD operations on resources. The main methods include:

GET: Retrieves data from the server without modifying it. It is safe and idempotent.

POST: Sends data to the server to create a new resource. It is not idempotent.

PUT: Updates an existing resource or creates it if it does not exist. It is idempotent.

DELETE: Removes a resource from the server. It is idempotent.

PATCH: Partially updates an existing resource. It is not necessarily idempotent.

These methods allow clients to interact with resources in a standardized way, supporting the core CRUD operations: Create (POST), Read (GET), Update (PUT/PATCH), and Delete (DELETE). Using these HTTP methods correctly ensures RESTful APIs are predictable, scalable, and easy to maintain.

Question 4: What is the purpose of the @app.route() decorator in Flask?
Answer:
The @app.route() decorator in Flask is used to bind a function to a specific URL route. It defines the endpoint at which a particular function will be executed when a client makes a request to that URL. This decorator allows Flask to map URLs to Python functions, enabling the creation of web routes that handle HTTP requests such as GET or POST. Essentially, it connects the web URL to the corresponding view function, facilitating the development of web applications by routing incoming requests to the appropriate handlers. This mechanism is fundamental in Flask for defining the behavior of different web pages or API endpoints within the application.

Additionally, the @app.route() decorator can specify the HTTP methods (like GET, POST) that the route will accept, providing control over how the endpoint responds to different types of requests. This flexibility is crucial for building RESTful APIs and dynamic web applications.


Question 5: What is the role of Flask-SQLAlchemy? (20 marks) Answer:

Flask-SQLAlchemy is an extension for Flask that adds support for SQLAlchemy, a powerful Object Relational Mapper (ORM) for Python. It simplifies database interactions by allowing developers to work with databases using Python classes and objects instead of writing raw SQL queries. Flask-SQLAlchemy provides a high-level abstraction for managing database connections, defining models, and performing CRUD operations.

The role of Flask-SQLAlchemy includes:

Simplifying database setup and configuration within Flask applications.

Enabling developers to define database schemas as Python classes (models).

Providing an easy-to-use API for querying and manipulating database records.

Supporting multiple database backends such as SQLite, MySQL, and PostgreSQL.

Integrating seamlessly with Flask’s application context and request lifecycle.

Overall, Flask-SQLAlchemy helps streamline database management in Flask apps, making it easier to build scalable and maintainable web applications.

Question 6: How do you create a basic Flask application? (20 marks) Answer:

To create a basic Flask application, follow these steps:

Install Flask using pip:

pip install Flask

Create a Python file (e.g., app.py) and import Flask:

from flask import Flask
app = Flask(__name__)

Define a route using the @app.route() decorator:

@app.route('/')
def home():
    return "Hello, Flask!"

Run the application:

if __name__ == '__main__':
    app.run(debug=True)

This basic app listens for requests at the root URL ('/') and returns a simple greeting. Running the app starts a local development server accessible via http://localhost:5000.

Question 7: How do you return JSON responses in Flask? (20 marks) Answer:

In Flask, you can return JSON responses using the jsonify function from the flask module. This function converts Python dictionaries or lists into JSON format and sets the appropriate content-type header.

Example:

from flask import Flask, jsonify
app = Flask(__name__)

@app.route('/data')
def data():
    response_data = {'name': 'John', 'age': 30}
    return jsonify(response_data)

This will return a JSON response with the data and the correct Content-Type: application/json header, making it easy for clients to consume the API response.

Question 8: How do you handle POST requests in Flask? (20 marks) Answer:

In Flask, POST requests are handled by defining a route that accepts the POST method and accessing the data sent in the request. Typically, you use the request object from the flask module to retrieve form data or JSON payloads.

Example:

from flask import Flask, request
app = Flask(__name__)

@app.route('/submit', methods=['POST'])
def submit():
    data = request.get_json()  # For JSON payload
    # Or use request.form for form data
    return f"Received data: {data}"

Question 9: How do you handle errors in Flask (e.g., 404)? (20 marks) Answer:

In Flask, errors such as 404 (Not Found) can be handled by defining custom error handlers using the errorhandler decorator. This allows you to create user-friendly error pages or responses when specific HTTP errors occur.

Example:

from flask import Flask, render_template
app = Flask(__name__)

@app.errorhandler(404)
def page_not_found(e):
    return render_template('404.html'), 404

This example shows how to catch 404 errors and return a custom HTML page with the appropriate status code. You can similarly handle other HTTP errors by defining additional error handlers.

Question 10: How do you structure a Flask app using Blueprints? (20 marks) Answer:

In Flask, Blueprints are a way to organize your application into modular components. They allow you to group related routes, templates, and static files, making your app more scalable and maintainable.

To structure a Flask app using Blueprints:

Create a Blueprint instance:

from flask import Blueprint

bp = Blueprint('bp', __name__)

Define routes within the Blueprint:

@bp.route('/example')
def example():
    return "This is a route in the blueprint"

Register the Blueprint with the main Flask app:

from flask import Flask
app = Flask(__name__)
app.register_blueprint(bp, url_prefix='/bp')

This setup allows you to keep your code organized by separating different parts of your application into Blueprints, each with its own routes and logic. The url_prefix parameter helps namespace the routes.

Using Blueprints is especially useful for larger applications or when building reusable components.

Question 10: How do you structure a Flask app using Blueprints?

Answer: In Flask, Blueprints are used to organize your application into modular components, making the codebase cleaner and easier to maintain. A Blueprint represents a collection of routes, templates, and static files that can be registered on the main Flask application. Structuring with Blueprints typically involves these steps:

Create a Blueprint object in a separate file (e.g., routes.py):
from flask import Blueprint

# Define the Blueprint
user_bp = Blueprint("user", __name__, url_prefix="/user")

# Add routes to the Blueprint
@user_bp.route("/profile")
def profile():
    return "User Profile Page"
Register the Blueprint in your main application file (e.g., app.py):

python
from flask import Flask
from routes import user_bp

app = Flask(__name__)

# Register the Blueprint
app.register_blueprint(user_bp)

if __name__ == "__main__":
    app.run(debug=True)
Organize your project structure for scalability:

Code
project/
│── app.py
│── routes/
│   └── user.py
│── templates/
│── static/
Advantages of Blueprints:

Promotes modularity by separating features into different files.

Makes large applications easier to maintain and extend.

Allows reusability of components across multiple projects.

Supports clean URL structures with url_prefix.
